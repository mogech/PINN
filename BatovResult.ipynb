## Постановка экспериментов

В ходе работы были исследованы различные конфигурации полносвязной нейронной сети, используемой для решения системы четырёх ОДУ.

В качестве основных параметров архитектуры рассматривались:

1. **Количество скрытых слоёв:** 3, 6 и 9.
2. **Количество нейронов в скрытом слое:** 32, 64 и 128.
3. **Функция активации:** tanh, sin, swish и ReLU.

Параметры модели и вид системы взяты из статьи:
**Dynamical Behaviors of Four Dimensional Prey-Predator Model**

2023
A. G. Farhan,
Nihad Sh. Khalaf,
Talat J. Aldhlki

Рассматривается следующая система из четырёх обыкновенных дифференциальных уравнений:

$$
\begin{aligned}
\frac{dx_1}{dt} &= x_1 \left( g_1 \left(1 - \frac{x_1}{k_1}\right) - \alpha_1 \frac{x_2 y_1}{x_1 + x_2} - \alpha_2 \frac{x_2 y_2}{x_1 + x_2} \right) \\[8pt]
\frac{dx_2}{dt} &= x_2 \left( g_2 \left(1 - \frac{x_2}{k_2}\right) - \beta_1 \frac{x_1 y_1}{x_1 + x_2} - \beta_2 \frac{x_1 y_2}{x_1 + x_2} \right) \\[8pt]
\frac{dy_1}{dt} &= y_1 \left( -\mu_1 + \varepsilon_1 \frac{x_1 x_2}{x_1 + x_2} - \rho_2 y_2 \right) \\[8pt]
\frac{dy_2}{dt} &= y_2 \left( -\mu_2 + \varepsilon_2 \frac{x_1 x_2}{x_1 + x_2} + \rho_1 y_1 \right)
\end{aligned}
$$

### Параметры модели

## Выводы

В ходе работы было исследовано влияние архитектуры нейронной сети и функции активации на качество решения системы 4 ОДУ методом PINN.

Основные результаты:

- Слишком узкие сети (например, 6×32) не справляются с задачей — невязка по $y_2$ перестаёт уменьшаться.
- Слишком широкие (6×128) и слишком глубокие (9×64) сети также дают плохой результат.
- Лучший результат показала архитектура **6 слоёв по 64 нейрона**.

При сравнении функций активации на архитектуре 6×64 выяснилось:
- `tanh` и `sin` работают хорошо;
- `swish` и `relu` приводят к такому же плохому результату, как и недостаточно ёмкие архитектуры
